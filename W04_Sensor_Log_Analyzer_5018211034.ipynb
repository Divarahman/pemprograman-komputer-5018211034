```python
sensor_logs = [i * 4 for i in range(1, 26)]
sensor_logs[2] = -5   # Sampel ke-3 sengaja dibuat negatif (tidak valid)
sensor_logs[7] = -15  # Sampel ke-8 sengaja dibuat negatif (tidak valid)
sensor_logs[14] = 92  # Sampel ke-15 melampaui batas (ALARM)
sensor_logs[20] = 85  # Sampel ke-21 melampaui batas (ALARM)

ambang_alarm = 75

total_nilai = 0
jumlah_valid = 0
jumlah_alarm = 0
nilai_min = None
nilai_max = None
nomor_sampel = 0

print(f"{'Sampel':<8} | {'Nilai':<6} | {'Status':<10}")
print("-" * 30)

# Loop eksplisit untuk memproses deretan data #
for nilai in sensor_logs:
    nomor_sampel += 1

# Mengabaikan nilai negatif sebagai data tidak valid #
    if nilai < 0:
        print(f"{nomor_sampel:<8} | {nilai:<6} | TIDAK VALID")
        continue

# Menentukan status NORMAL/ALARM #
    if nilai > ambang_alarm:
        status = "ALARM"
        jumlah_alarm += 1
    else:
        status = "NORMAL"

    print(f"{nomor_sampel:<8} | {nilai:<6} | {status}")

    total_nilai += nilai
    jumlah_valid += 1

# Mencari nilai minimum dan maksimum secara manual tanpa min()/max() #
    if nilai_min is None or nilai_max is None:
        nilai_min = nilai
        nilai_max = nilai
    else:
        if nilai < nilai_min:
            nilai_min = nilai
        if nilai > nilai_max:
            nilai_max = nilai

# Menghitung rata-rata secara manual tanpa sum() #
if jumlah_valid > 0:
    rata_rata = total_nilai / jumlah_valid
else:
    rata_rata = 0

print("\n" + "="*30)
print("HASIL ANALISIS SENSOR LOG")
print("="*30)
print(f"Rata-rata nilai valid : {rata_rata:.2f}")
print(f"Total kejadian alarm   : {jumlah_alarm}")
print(f"Nilai minimum valid    : {nilai_min}")
print(f"Nilai maksimum valid   : {nilai_max}")
```
